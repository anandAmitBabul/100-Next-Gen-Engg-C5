Chain 1 - creates business kpis from the financial dataset provided

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
# prompt for Chain 1
kpi_prompt=PromptTemplate(
    input_variables=["data"],
    template="""
    You are a financial analyst and you have been proided with
    a financial data. Extract the most important financial and business KPIs from the data
    and present them in bullet point format

    financial data:{data}"""
)
```

```python
chain1=kpi_prompt | llm | StrOutputParser()
```

```python
# executing chain 1
kpis=chain1.invoke({
    "data":data
})
```

```python
print(kpis)
```

5th November 2026

```python
benchmark_prompt=PromptTemplate(
    input_variables=["kpis"],
    template="""
    You are a Senior Strategy Consultant and your job is to critically analyze the kpis
    and establish appropriate performance benchmarks for management evaluation

    For every major KPI provide the following :
    - Define a reasonable target or benchmark
    - Define the current performance
    - Classify the performance as
        - Strong
        - Moderate
        - Needs Improvement

kpis:{kpis}"""
)
```

```python
chain2=benchmark_prompt | llm | StrOutputParser()
```

```python
# executing chain 2
benchmarks=chain2.invoke({
    "kpis":kpis
})
```

```python
print(benchmarks)
```

Activity - Create chain3 that allows us to generate an executive report by taking into account the output of chain2 which is the benchmarks

```python
exec_report_prompt=PromptTemplate(
    input_variables=["benchmarks"],
    template="""
    You are a Senior Executive and your job is to analyze the benchmarks
    and generate executive report

    You must read all the benchmarks from the input :
    - Define benchmark in a way how it works
    - Define the benchmark on how impactful it is
    - Classify the performance in 3 groups based upon their impact

benchmarks:{benchmarks}"""
)
```

```python
chain3=exec_report_prompt | llm | StrOutputParser()
```

```python
# executing chain 2
report=chain3.invoke({
    "benchmarks":benchmarks
})
```

```python
print(report)
```
