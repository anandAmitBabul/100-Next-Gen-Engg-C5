Chain 1 - creates business kpis from the financial dataset provided

```python
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
# prompt for Chain 1
kpi_prompt=PromptTemplate(
    input_variables=["data"],
    template="""
    You are a financial analyst and you have been proided with
    a financial data. Extract the most important financial and business KPIs from the data
    and present them in bullet point format

    financial data:{data}"""
)
```

```python
chain1=kpi_prompt | llm | StrOutputParser()
```

```python
# executing chain 1
kpis=chain1.invoke({
    "data":data
})
```
