```python
from langchain_community.utilities import SerpAPIWrapper
from langchain_core.tools import tool
search_client=SerpAPIWrapper()
```

Whenever we create tols , we ensure thatthe following are incorporated:

*   Name of the tool
*   Dexcription of the tool
*   Input schema of the tool

```python
@tool
def web_search(query:str)->str:
    """Searches the live web using google search results,
    it then uses the tool to retrieve the current information,
    company announcements, industry developments, market reports
    and any other supporting evidence"""
    return search_client.results(query)

# Example usage storing the result in 'results'
results = web_search.invoke({"query": "your search query"})
```

```python
# creating the shared memory - state
from typing import TypedDict

class ResearchState(TypedDict):
  user_query:str
  research_findings:str
  strategic_analysis:str
  report:str
```

```python
# creating the first node - agent 1 - research agent
from langchain_core.prompts import PromptTemplate

research_prompt=PromptTemplate(
    input_variables=["user_query"],
    template="""
    You are a business research analyst ad your responsiibility is to
    investigate business problems using live web search.

    Use the web_search tool to retrieve the information from the internet
    and do not invent statistics or announcements or any other market developments

    Follow the format as mentioned below:
    - Research Findings
    - Industry developments
    - Company development (if relevant)
    - Opportunities
    - Challenges and Risks
""")
```
